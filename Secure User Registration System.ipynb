    The email must contain both "@" and ".".

Create an orchestrator function called validate_user_data(name, email, password)

This function must:
<ul>

<li>Raise a ValueError with a clear and descriptive message if any validation fails</li>
<li>Return True if all validations pass successfully</li>
</ul>


Part 4: Create the Registration Function
===================================================

Create a function called create_user_account(name, email, password)

This function must:

<ul>

        Catch the ValueError.
        <br/>Store a dictionary inside failed_registrations containing the email and the error message.
        <br/>Return None.
</ul>

```python
registered_users = [] #To store successfully registered users
failed_registrations = [] #To store information about failed registration attempts
```

```python
def validate_name(name:str)-> bool:
     '''
        checks username meets minimum requirement of 3 characters

        Args : 
            name (str) : User name
        Returns:
            True: if successful False: otherwise         
    '''
    
    return len(name) >= 3
```

```python
def validate_email(email:str)-> bool:
    '''
        checks email contains . and @ sign

        Args : 
            email (str) : email
        Returns:
            True: if successful False: otherwise         
    '''
    
    return "." in email and "@" in email
```

```python
def validate_password(password:str)-> bool:
    '''
        Checks password meet all of the following conditions:

        At least 8 characters long
        Contains at least one uppercase letter
        Contains at least one digit
        
        Args : 
            password (str) : password
        Returns:
            True: if successful False: otherwise         
    '''
    
    has_valid_length = len(password) >= 8

    has_digit = any(char.isdigit() for char in password)
    
    has_uppercase = any(char.isupper() for char in password)
      
    return has_valid_length and has_digit and  has_uppercase
```

```python
def validate_user_data(name:str, email:str, password:str)->bool:
    
    '''
        This is a controller/Orchestrator function that takes name,email,password as an input and process it
        Args : 
            name (str) : User name
            email  (str) : User email
            password (str) : User password

        Returns:
            True: if all three validations are successful          
    '''
    if not validate_name(name):
        raise ValueError(email,f"Invalid name: {name}. The name must contain at least 3 characters.")
    if not validate_email(email):
        raise ValueError(email,f"Invalid email: {email}. The email must contain both '@' and '.'")
    if not validate_password(password):
        raise ValueError(email,f"Invalid password: {password}.The password must meet all of the following conditions:At least 8 characters long,Contains at least one uppercase letter,Contains at least one digit")
    
    return True
    
```

```python
def create_user_account(name, email, password):
    '''
        This is a user registration function that calls Orchestrator function, flags if duplicate email is used for registration.
        Valid user regisration data is saved in registered_users list.
        Args : 
            name (str) : User name
            email  (str) : User email
            password (str) : User password

        Returns:
            Dictionary of new user registration details : if successful   
            None : otherwise
    '''
   
    validate_user_data(name,email,password)

    if any(email in d.values() for d in registered_users):
        raise ValueError(email,"The email already registered")
        return None
        
    d = {"name" : name, "email" : email, "password" : password, "status" : "Active"}
    registered_users.append(d)
    return d
```
