It allows users to record their spending, organize expense details, monitor spending habits, and calculate total expenses. The main objective of this project is to make expense management easier, improve financial awareness, and help users maintain better control over their budgets.

```python
expenses = [] #This list will store all expense entries.
```

```python
def add_expense(amount:float, category:str, description:str)->dict: 
    '''
        Checks amount is greater than zero. If not raise an exception.
        Create a dictionary with an expense detail and add to the expense list.       

        Args : 
            amount(float): expense amount
            category(str): expense category
            decription(str): description about category
        Returns:
            dictionary of exepnse details: if successful          
    '''
    if amount <= 0:
        raise ValueError(f"Amount for {category} must be greater than zero.")
    d = {'category':category, 'description': description,'amount':float(amount)}
    expenses.append(d)
    return d
```

```python
def calculate_total_expenses()->float:
    '''
        Calculates the sum of all expenses   
        Returns: sum of all expenses 
    '''
    
    total_expenses = sum(d['amount'] for d in expenses)
    return total_expenses   
 
```

```python
def calculate_total_by_category(category:str)->float:
    '''
        Calculates the sum of expenses for the given category 
        
        Args:
            category(str) : exepnse category
            
        Returns: sum of expenses for the given category
    '''
    
    total_expenses_category = sum(d['amount'] for d in expenses if d['category'] == category)
    return total_expenses_category    
```

```python
def show_expenses(): 
    '''
        Displays all expenses details
        
    '''
    for expense in expenses:
        print(f"{expense['category']}: ${expense['amount']:.2f} - {expense['description']}")
```

```python
#--------------------Testing and exception handling------------------------
try:
    #add_expense(amount = 0, category = "Food", description = "Lunch at cafe")
    add_expense(amount = 24.50, category = "Food", description = "Lunch at cafe")
    add_expense(amount = 45.00, category = "Transport", description = "Gas refill")
    add_expense(amount = 15.99, category = "Entertainment", description = "Movie ticket")
    add_expense(amount = 60.00, category = "Utilities", description = "Internet bill")
    add_expense(amount = 200.00, category = "Utilities", description = "Water bill")
    
    print(f"Total Expenses:${calculate_total_expenses():.2f}")

    category = "Utilities"
    print(f"Total Expenses for {category}:${calculate_total_by_category(category):.2f}")
    

except ValueError as e:
    print(e)

finally:
    print("------------List of all expenses--------")
    show_expenses()
```
