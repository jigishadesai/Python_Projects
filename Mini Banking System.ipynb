```python
accounts = [] #This list will store all bank accounts
```

```python
def create_account(name:str, initial_balance:float):
    
    if initial_balance <= 0:
        raise ValueError(f"Initial balance amount for '{name}' must be greater than zero.")
        
    if any(name in d.values() for d in accounts):
        raise ValueError(f"The account with '{name}' has already been registered.")

    d = {'name':name,'balance': float(initial_balance), 'transactions':[{'type':'Deposit', 'amount':float(initial_balance)}]}
    accounts.append(d)
    return d    
```

```python
def deposit(name:str, amount:float):
     
    account = next ( (acc for acc in accounts if acc['name'] == name), None)

    if account is None:
        raise ValueError(f"Account for '{name}' not found.")
    
    if amount <= 0:
        raise ValueError(f"Deposit Amount for '{name}' must be greater than zero.")

    amount = float(amount)
    
    account['balance'] += amount

    account['transactions'].append({
    'type': 'Deposit',
    'amount': amount
    })

    return account['balance']
```

```python
def withdraw(name:str, amount:float):

    account = next((acc for acc in accounts if acc['name'] == name),None)

    if account is None:
        raise ValueError(f"Account for '{name}' not found.")

    if amount <= 0:
        raise ValueError(f"Withdrawal Amount for '{name}' must be greater than zero.")
    
    amount = float(amount)

    if account['balance'] < amount:
        raise ValueError(f"Insufficient funds in {name}'s account.")

    account['balance'] -= amount

    account['transactions'].append({
    'type': 'Withdrawal',
    'amount': amount
    })

    return account['balance']
   
```

```python
def show_account(name:str):

    account = next((acc for acc in accounts if acc['name'] == name), None)

    if account is None:
        raise ValueError(f"Account for '{name}' not found.")

    print("-------------------------------")
    print(f"Account holder name: {account['name']}")
    print(f"Account Balance: ${account['balance']:.2f}")
    print("---------Transactions----------")

    for tr in account['transactions']:
        print(f"{tr['type']}: ${tr['amount']:.2f}")
 
```
